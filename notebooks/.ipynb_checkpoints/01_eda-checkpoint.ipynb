```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load_data, basic_cleaning
from src.eda_utils import plot_loss_ratio_by_province, plot_premium_vs_claims

sns.set_style("whitegrid")
%matplotlib inline

# Load data
df = load_data()
df = basic_cleaning(df)
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nData Types:\n", df.dtypes)
print("\nDescriptive Statistics:\n", df.describe())

# Missing values
missing = df.isnull().sum() / len(df) * 100
print("\nMissing Values (%):\n", missing[missing > 0].sort_values(ascending=False))
```

```python
# Example handling strategy
# df = df.dropna(subset=['TotalPremium', 'TotalClaims'])  # or impute
print("Missing value handling strategy documented.")
```

```python
# 1. Loss Ratio by Province
plot_loss_ratio_by_province(df)

# 2. Premium vs Claims
plot_premium_vs_claims(df)

# 3. Boxplot for Outliers
plt.figure(figsize=(10,6))
sns.boxplot(data=df, y='TotalClaims')
plt.title('Outliers in Total Claims')
plt.show()

# Additional good one
sns.countplot(data=df, x='Gender')
plt.title('Policies by Gender')
plt.show()
```
