```python
# ================================================
# 02_hypothesis_testing.ipynb
# ================================================

# === FIX PATH ===
import sys
import os
sys.path.append(os.getcwd())
sys.path.append(os.path.join(os.getcwd(), 'src'))
sys.path.append(os.path.join(os.getcwd(), '..'))

import pandas as pd

# Import modules
from data_loader import load_data, basic_cleaning
from hypothesis_tests import calculate_claim_frequency, test_provinces, test_gender_risk, test_zipcode_margin

# ====================== LOAD DATA ======================
print("="*70)
print("LOADING DATA FOR HYPOTHESIS TESTING")
print("="*70)

df = load_data()
df = basic_cleaning(df)
df = calculate_claim_frequency(df)

print(f"Dataset ready for testing. Shape: {df.shape}")

# ====================== RUN TESTS ======================
print("\nRunning Statistical Tests...\n")

results = []
results.append(test_provinces(df))
results.append(test_gender_risk(df))
results.append(test_zipcode_margin(df))

# Create results table
results_df = pd.DataFrame([r for r in results if r is not None])
print("=== Hypothesis Testing Results ===")
print(results_df)

# Save results (optional)
results_df.to_csv("reports/hypothesis_results.csv", index=False)
print("\n✅ Hypothesis Testing Completed!")
```
